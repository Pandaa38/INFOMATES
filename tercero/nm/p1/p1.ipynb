```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir la malla de puntos en el cuadrado [-2, 2] x [-2, 2]
# Usamos 1000 puntos para que las curvas se vean muy suaves y precisas
x = np.linspace(-2,-2, 1000)
y = np.linspace(-2, 2, 1000)
X, Y = np.meshgrid(x, y)

# 2. Definir las funciones f(x,y) y g(x,y)
# Nota: Escribimos X e Y en mayúsculas porque operamos sobre las matrices completas
F = 2*X**5 - 20*X**3*Y**2 - 4*X**3*Y + 10*X*Y**4 + 4*X*Y**3 - 2*X
G = 10*X**4*Y + X**4 - 20*X**2*Y**3 - 6*X**2*Y**2 + 2*Y**5 + Y**4 - 2*Y - 1

# 3. Configurar el tamaño del gráfico
plt.figure(figsize=(10, 10))

# 4. Dibujar las curvas implícitas igualadas a 0 (levels=[0])
contour_f = plt.contour(X, Y, F, levels=[0], colors='blue')
contour_g = plt.contour(X, Y, G, levels=[0], colors='red')

# 5. Configurar leyendas (creamos unas líneas invisibles para referenciarlas)
plt.plot([], [], color='blue', label='$f(x,y)=0$')
plt.plot([], [], color='red', label='$g(x,y)=0$')

# 6. Detalles estéticos del gráfico
plt.title('Intersección de las curvas implícitas (5 Soluciones)', fontsize=14)
plt.xlabel('Eje X', fontsize=12)
plt.ylabel('Eje Y', fontsize=12)

# Destacamos las 5 soluciones exactas con puntos verdes
soluciones_x = [1, -1, 0, 0, 0]
soluciones_y = [0, 0, 1, -1, -0.5]
plt.plot(soluciones_x, soluciones_y, 'go', markersize=8, label='Soluciones del sistema', zorder=5)

# Poner una cuadrícula de fondo
plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.7)

# Límites del cuadrado
plt.xlim([-2, 2])
plt.ylim([-2, 2])

plt.legend(fontsize=12)

# Mostrar el gráfico
plt.show()
```
